```python
!pip  install -q transformers
from transformers import pipeline
classifier = pipeline("zero-shot-classification", model = "facebook/bart-large-mnli")
print ("intent classifier ready.")

```

#Define your intent catagries

```python
intent_labels =[
    "booking a table",
    "cancelling an order",
    "asking about pricing",
    "making a complaint",
    "general greeting",
]

result = classifier("I would like to reserve a table for two at 8 PM", intent_labels)
print("Message:", result["sequence"])
for label, score in zip(result["labels"], result["scores"]):
    print(f"{label:25s}{score:.3f}")
```

```python
intent_labels =[
    "booking a table",
    "cancelling an order",
    "asking about pricing",
    "making a complaint",
    "general greeting",
]

result = classifier("whoud you give any complaint", intent_labels)
print("Message:", result["sequence"])
for label, score in zip(result["labels"], result["scores"]):
    print(f"{label:25s}{score:.3f}")
```

```python
test_messages = [
    "This is the third time my order hasn't arrived, I am really upset",
    "Hi! Good morning",
    "How much do you charge for a large pizza?",
    "I need to cancel my order number 4521",
]

for msg in test_messages:
    result = classifier(msg, intent_labels)
    top_intent = result["labels"][0]
    top_score = result["scores"][0]
    print(f'"{msg}"')
    print(f" -> {top_intent} (confidence {top_score:.2f})\n")
```
